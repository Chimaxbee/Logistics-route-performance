**LOGISTICS ROUTE PERFORMANCE REPORT**


```python
delivery_vehicle = "Truck A12"
delivery_route = "Lagos to Ibadan"
planned_distance = 135
actual_distance_covered = 128
fuel_used = 26
fuel_efficiency_target = 5.0
additional_distance = 12
maximum_allowed_distance = 145
vehicle_available = True
```

```python
#Task 2 - Calculate Route Difference
route_difference = planned_distance - actual_distance_covered
# Result
print(f"Route Difference: {route_difference} km")
```

```python
#Task 3 Calculate Fuel Efficiency
fuel_efficiency = actual_distance_covered / fuel_used

# Display the result rounded to two decimal places.
print(f"Fuel Efficiency: {fuel_efficiency:.2f} km/litre")
```

```python
# Task 4 - Update Distance

# actual distance covered due to the diversion.
actual_distance_covered += additional_distance

# Display the updated distance.
print(f"Updated Distance Covered: {actual_distance_covered} km")
```

```python
# Task 5 - Within Allowed Distance

# Distance within the maximum allowed distance.
within_allowed_distance = actual_distance_covered <= maximum_allowed_distance

# result.
print(f"The vehicle stayed within Allowed Distance: {within_allowed_distance}")
```

```python
# Task 6 - Fuel Target Achieved

# Check if the fuel efficiency target was achieved.
fuel_target_achieved = fuel_efficiency >= fuel_efficiency_target

# result.
print(f"The efficiency target was not achieved: {fuel_target_achieved}")
```

```python
# Task 7 - Route Efficient
# Determine if route is efficient.
route_efficient = vehicle_available and within_allowed_distance
# result.
print(f"The Route is Efficient: {route_efficient}")
```

```python
# Task 8 - Operational Review
# Determine whether an operational review is required.
operational_review_required = (not route_efficient) or (not fuel_target_achieved)
#result.
print(f"Operational Review is Required: {operational_review_required}")
```

```python

# Task 9 - Professional Logistics Report

print(f"""
=========================================================
          LOGISTICS ROUTE PERFORMANCE REPORT
=========================================================

Delivery Vehicle: {delivery_vehicle}
Delivery Route: {delivery_route}
Planned Distance: {planned_distance} km
Updated Distance Covered: {actual_distance_covered} km
Distance Difference: {route_difference} km
Fuel Used: {fuel_used} litres
Fuel Efficiency: {fuel_efficiency:.2f} km/litre
Fuel Efficiency Target: {fuel_efficiency_target:.2f} km/litre

Maximum Allowed Distance: {maximum_allowed_distance} km
Within Allowed Distance: {within_allowed_distance}
Fuel Efficiency Target Achieved: {fuel_target_achieved}
Vehicle Available: {vehicle_available}
Route Efficient: {route_efficient}
Operational Review Required: {operational_review_required}

=========================================================
""")
```
